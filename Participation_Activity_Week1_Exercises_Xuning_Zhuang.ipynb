## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
manualmean = sum(salaries) / len(salaries)

import statistics

verifiedmean = statistics.mean(salaries)

print("Manual mean:", manualmean)
print("Verified mean:", verifiedmean)
```

### median

```python
sortedsalaries = sorted(salaries)
numberofsalaries = len(sortedsalaries)

middleright = numberofsalaries // 2
middleleft = middleright - 1
manualmedian = (sortedsalaries[middleright] + sortedsalaries[middleleft]) / 2

verifiedmedian = statistics.median(salaries)

print("Manual median:", manualmedian)
print("Verified median:", verifiedmedian)

# I wrtite this code by using chatgpt, bucause I am not sure how to translate the math expression of the median to the coding format.
```

### mode

```python
frequency = {}

for salary in salaries:
    if salary in frequency:
        frequency[salary] = frequency[salary] + 1
    else:
        frequency[salary] = 1

manualmode = None
highestfrequency = 0

for salary, count in frequency.items():
    if count > highestfrequency:
        manualmode = salary
        highestfrequency = count

verifiedmode = statistics.mode(salaries)

print("Manual mode:", manualmode)
print("The mode appears", highestfrequency, "times")
print("Verified mode:", verifiedmode)

# I use AI to help me to make sure whether my For loop are written in a right way
```

### sample variance
Remember to use Bessel's correction.

```python
squareddifferences = []

for salary in salaries:
    difference = salary - manualmean
    squareddifferences.append(difference ** 2)

manualvariance = sum(squareddifferences) / (len(salaries) - 1)
verifiedvariance = statistics.variance(salaries)

print("Manual sample variance:", manualvariance)
print("Verified sample variance:", verifiedvariance)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
manualsd = manualvariance ** 0.5
verifiedsd = statistics.stdev(salaries)

print("Manual sample standard deviation:", manualsd)
print("Verified sample standard deviation:", verifiedsd)
```

## Exercise 6: Calculating more statistics
### range

```python
salaryrange = max(salaries) - min(salaries)

print("Range:", salaryrange)
```

### coefficient of variation

```python
coeff_ofvar = manualsd / manualmean

print("Coefficient of variation:", coeff_ofvar)
print("Coefficient of variation as a percentage:", coeff_ofvar * 100, "%")
```

### interquartile range

```python
quartiles = statistics.quantiles(salaries, n=4, method="inclusive")
q1 = quartiles[0]
q3 = quartiles[2]

interquartilerange = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("Interquartile range:", interquartilerange)
```

### quartile coefficent of dispersion

```python
quartilecoeff_ofdispersion = interquartilerange / (q3 + q1)

print("Quartile coefficient of dispersion:", quartilecoeff_ofdispersion)
```

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaledsalaries = [
    (salary - min(salaries)) / salaryrange
    for salary in salaries
]

print("First 10 min-max scaled salaries:", min_max_scaledsalaries[:10])
print("Minimum scaled value:", min(min_max_scaledsalaries))
print("Maximum scaled value:", max(min_max_scaledsalaries))
```

### standardizing

```python
standardized_salaries = [
    (salary - manualmean) / manualsd
    for salary in salaries
]

print("First 10 standardized salaries:", standardized_salaries[:10])
print("Mean after standardizing:", statistics.mean(standardized_salaries))
print("Sample standard deviation after standardizing:", statistics.stdev(standardized_salaries))
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
# I use AI to create these codes. It remind me that the covariance of one same variable equals to the variance of that variabl itself.

covariance_products = [
    (salary - manualmean) * (salary - manualmean)
    for salary in salaries
]
manualcovariance = sum(covariance_products) / (len(salaries) - 1)
verifiedcovariance = statistics.covariance(salaries, salaries)

print("Manual sample covariance:", manualcovariance)
print("Verified sample covariance:", verifiedcovariance)
```

### Pearson correlation coefficient ($\rho$)

```python
manualcorr = manualcovariance / (manualsd * manualsd)
verifiedcorr = statistics.correlation(salaries, salaries)

print("Manual Pearson correlation coefficient:", manualcorr)
print("Verified Pearson correlation coefficient:", verifiedcorr)
```
